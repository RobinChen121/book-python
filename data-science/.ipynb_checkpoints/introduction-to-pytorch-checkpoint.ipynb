## Tensor
<hr>

In deep learning, tensors are the fundamental building blocks of data. As multidimensional arrays of numbers, they generalize scalars, vectors, and matrices. 

- 0D Tensor: a scalar (a single number, like 5).

- 1D Tensor: a vector (a list of numbers, like [1, 2, 3]).

- 2D Tensor: a matrix (a table of numbers with rows and columns).

- 3D+ Tensors: data with more dimensions (like a color image with Height, Width, and Color Channels).

While similar to NumPy arrays, tensors are specifically designed to support automatic differentiation and GPU acceleration.

- Automatic differentiation works like a smart calculator that remembers every math step performed during a computation. As tensors go through operations such as addition or multiplication, PyTorch keeps track of how each result was produced. Later, when training a neural network, PyTorch goes backward through these recorded steps to figure out how much each tensor contributed to the final result. These contributions, called gradients, are automatically calculated and stored in the tensor’s `.grad` field, so the model knows how to adjust its parameters to improve performance.

- Deep learning involves millions of small mathematical operations, such as multiplying numbers together. Without a GPU (Graphics Processing Unit), the CPU (Central Processing Unit) must handle these computations alone. Although modern CPUs are fast, processing millions of operations still takes considerable time. GPUs are specifically designed to perform many of these small computations simultaneously. As a result, a task that might take a CPU an hour could be completed by a GPU in just a few seconds.

### Creation
<hr>

The following functions are used to initialize tensors from scratch or existing data.

- `torch.tensor()`: creates a tensor from data (like a list, tuple or NumPy array).

`````python
import torch

data = [[1, 2, 3], [4.2, 5 ,6]]

torch_data1 = torch.tensor(data) # from a list
print(torch_data1)
print(torch_data1.dtype) # get the data type of a tensor
`````

`````python
import numpy as np

np_data = np.ones((2, 3))
torch_data2 = torch.tensor(np_data)
print(torch_data2)
`````

- Create with random or sequence values by `torch.zeros()`, `torch.ones()`, `torch.randn()`, `torch.arange()` or `torch.linspace()`.

`````python
shape = (3, 2)
torch.zeros(shape)
`````

`````python
torch.ones((1,3))
`````

`````python
torch.randn((2,1)) # returns a tensor filled with random numbers from a normal distribution (mean 0, variance 1)
`````

`````python
torch.arange(2, 20, 2) # torch.arange(start, end(exclusive), step) returns a 1D tensor with a range of values
`````

`````python
torch.linspace(2, 20, 3) # torch.linspace() torch.linspace(start, end(inclusive), steps) returns a 1D tensor of steps equally spaced points between start and end
`````

```{note}
When converting a NumPy array to a PyTorch Tensor, PyTorch inherits NumPy’s default floating-point precision *float64*. However, it is better to use the data type *float32* in deep learning for faster computation. 
```

- Use the `.float()` method to cast a tensor to the float32 data type for later training.

`````python
x = torch.arange(10)
print(x.dtype)
y = x.float()
print(y.dtype)
`````

### Shape manipulation
<hr>

PyTorch has many shape manipulation functions because deep learning models frequently require tensors to be reshaped, transposed, or expanded between layers with different input requirements.

- View the shape of a tensor: `.shape`, `.ndim`.

`````python
x = torch.ones((3,4)) 
x.shape # another function .size() return same result as .shape
`````

`````python
x.ndim # get the dimension of a tensor
`````

. Change the shape of a tensor: `torch.squeeze()`, `.unsqueeze()`, `.flatten()`.

`torch.squeeze()` is to squeeze a tensor by removing the dimensions of size 1. Its syntex is:

<table>   
    <tr style="border-top:solid; border-bottom:solid">
        <th colspan=2 style="text-align:center">
            torch.squeeze(input, dim=None)
        </th>
    </tr>
    <tr>
        <td style="text-align:left">input</td>
        <td style="text-align:left">The input tensor.</td>
    </tr>
    <tr style="border-bottom:solid">
        <td style="text-align:left">dim</td>
        <td style="text-align:left">Given dimensions to squeeze.</td>
    </tr>
</table>

`````python
x = torch.ones((2, 1, 3, 1, 4))
x.shape
`````

`````python
y = torch.squeeze(x) # remove all the dimensions of size 1
y.shape
`````

`````python
y = torch.squeeze(x, 0)
y.shape # the shape does not change as the 0 dimension is not size 1
`````

`````python
y = torch.squeeze(x, dim=(1, 2)) # squeeze the size 1 dimension among the given dimensions
y.shape
`````

- `unsqueeze()` is used to add a new dimension of size 1 at the specified position to a tensor.

<table>   
    <tr style="border-top:solid; border-bottom:solid">
        <th colspan=2 style="text-align:center">
            torch.unsqueeze(input, dim)
        </th>
    </tr>
    <tr>
        <td style="text-align:left">input</td>
        <td style="text-align:left">The input tensor.</td>
    </tr>
    <tr style="border-bottom:solid">
        <td style="text-align:left">dim</td>
        <td style="text-align:left">Given dimension (should be lower equal then the number of dimensions of the tensor) to unsqueeze.</td>
    </tr>
</table>

`````python
x = torch.tensor([1, 2, 3])
print(x)
x.shape
`````

`````python
y = torch.unsqueeze(x, 0)
print(y)
y.shape
`````

`````python
y = torch.unsqueeze(x, 1)
print(y)
y.shape
`````

`````python
x = torch.tensor([[1, 2, 3], [4, 5, 6]])
print(x.shape)
y = torch.unsqueeze(x, 0)
print(y)
print(y.shape)
`````

`````python
x = torch.tensor([[1, 2, 3], [4, 5, 6]])
print(x.shape)
y = torch.unsqueeze(x, 1)
print(y)
print(y.shape)
`````

`````python
x = torch.tensor([[1, 2, 3], [4, 5, 6]])
print(x.shape)
y = torch.unsqueeze(x, 2)
print(y)
print(y.shape)
`````

- `torch.flattern()` is to reshape a tensor to be one-dimensional.

`````python
x = torch.tensor([[1, 2, 3], [5, 6, 7]])
print(x)
print(x.shape)
y = torch.flatten(x)
print(y.shape)
print(y)
`````

### Concanetation
<hr>

- `torch.cat()` is to concanetate tensors at a given dimension.

<table>   
    <tr style="border-top:solid; border-bottom:solid">
        <th colspan=2 style="text-align:center">
            torch.cat(tensors, dim=0)
        </th>
    </tr>
    <tr>
        <td style="text-align:left">tesors</td>
        <td style="text-align:left">A sequence of tensors, which must have the same shape, except in the cat dimension.</td>
    </tr>
    <tr style="border-bottom:solid">
        <td style="text-align:left">dim</td>
        <td style="text-align:left">A scalar. The given dimension to concanetate.</td>
    </tr>
</table>

`````python
a = torch.tensor([[1, 2], [3, 4]])
b = torch.tensor([[5, 6], [7, 8]])
print(a)
print(b)
c = torch.cat((a, b), dim=0)
print(c)
print(c.shape)
`````

`````python
a = torch.tensor([[1, 2], [3, 4]])
b = torch.tensor([[5, 6], [7, 8]])
print(a)
print(b)
c = torch.cat([a, b], dim=1)
print(c)
print(c.shape)
`````

- `torch.stack()` is to concanetate tensors along with a **new** dimension.

<table>   
    <tr style="border-top:solid; border-bottom:solid">
        <th colspan=2 style="text-align:center">
            torch.stack(tensors, dim=0)
        </th>
    </tr>
    <tr>
        <td style="text-align:left">tesors</td>
        <td style="text-align:left">A sequence of tensors to concanetate and the tensors need to be of the same size.</td>
    </tr>
    <tr style="border-bottom:solid">
        <td style="text-align:left">dim</td>
        <td style="text-align:left">The given dimension to concanetate and shoule be between 0 and the number of dimensions of concatenated tensors (inclusive).</td>
    </tr>
</table>

````{note}
If each tensor has shape (m, n) and there are k tensors to stack, the resulting shape is (k, m, n) for `dim=0`, (m, k, n) for `dim=1`, and (m, n, k) for `dim=2`. More generally, `dim=d` inserts a new dimension at position d, whose size is k.
````

`````python
a = torch.tensor([[1, 2, 3], [4, 5, 6]])
b = torch.tensor([[7, 8, 9], [10, 11, 12]])
print(a.shape)
c = torch.stack([a, b], dim=0) # stack a, b along dimension 0 to [a, b]
print(c)
print(c.shape)
print(c[0])
`````

`````python
a = torch.tensor([[1, 2, 3], [4, 5, 6]])
b = torch.tensor([[7, 8, 9], [10, 11, 12]])

c = torch.stack([a, b], dim=1) # stack a, b along dimension 1
print(c)
print(c.shape)
print(c[0][0])
print(c[1][0])
`````

`````python
c = torch.stack([a, b], dim=2) # stack a, b along dimension 2
print(c)
print(c.shape)
print(c[0])
`````

````{note}
Literally, `torch.stack(tensors, dim=d)` first unsqueeze each teansor along the given dimension d and then concanetate them at the given dimension d.
````

`````python
a = torch.tensor([[1, 2, 3], [4, 5, 6]])
x = torch.unsqueeze(a, 1)
print(x)
`````

`````python
b = torch.tensor([[7, 8, 9], [10, 11, 12]])
y = torch.unsqueeze(b, 1)
print(y)
`````

`````python
torch.cat([x, y], dim=1)
`````

`````python
torch.cat([x, y], dim=1)
`````

`````python
a = torch.tensor([[1, 2, 3], [4, 5, 6]])
x = torch.unsqueeze(a, 0)
print(x)
`````

`````python
b = torch.tensor([[7, 8, 9], [10, 11, 12]])
y = torch.unsqueeze(b, 0)
print(y)
`````

`````python
torch.cat([x, y], dim=0)
`````
